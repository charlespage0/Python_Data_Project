```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

#loading data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
top_countries = df['job_country'].value_counts().sort_values(ascending=False).head(6).index
main_titles = ['Data Analyst', 'Data Engineer', 'Data Scientist']
df_country_table = df.pivot_table(values = 'salary_year_avg',index='job_country', columns = 'job_title_short', aggfunc='median')
df_country_table.loc[top_countries, main_titles].plot(kind = 'bar')
plt.xticks(rotation=45)
```
